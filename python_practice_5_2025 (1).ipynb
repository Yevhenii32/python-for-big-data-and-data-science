6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math

class Cube(figure):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side ** 3

class Cylinder(figure):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * (self.radius ** 2) * self.height

my_cube = Cube(3)
my_cylinder = Cylinder(2, 5)

total_volume = my_cube + my_cylinder

print(f"Об'єм куба: {my_cube.get_volume():.2f}")
print(f"Об'єм циліндра: {my_cylinder.get_volume():.2f}")
print(f"Сумарний об'єм: {total_volume:.2f}")
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, wallet_id, wallet_balance):
        self.__id = wallet_id
        self.__wallet_balance = wallet_balance

    @property
    def balance(self):
        return self.__wallet_balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            print("Баланс не може бути від'ємним!")
        else:
            self.__wallet_balance = amount

    @property
    def wallet_id(self):
        return self.__id

    def addition(self, amount):
        if amount > 0:
            self.__wallet_balance += amount
            print(f"Рахунок поповнено на {amount}. Поточний баланс: {self.__wallet_balance}")
        else:
            print("Сума поповнення має бути більшою за 0")

    def withdraw(self, amount):
        if amount > self.__wallet_balance:
            print("Недостатньо коштів!")
        elif amount <= 0:
            print("Сума має бути більшою за 0")
        else:
            self.__wallet_balance -= amount
            print(f"Знято {amount}. Поточний баланс: {self.__wallet_balance}")


wallet = EWallet("EW123", 1000)

print("ID гаманця:", wallet.wallet_id)
print("Баланс:", wallet.balance)

wallet.addition(400)
wallet.withdraw(300)
wallet.balance = -200
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price, quantity=1):
        self.name = name
        self.price = price
        self.quantity = quantity


class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)
        print(f"{item.name} додано до кошика.")

    def remove_item(self, item_name):
        for item in self.items:
            if item.name == item_name:
                self.items.remove(item)
                print(f"{item_name} видалено з кошика.")
                return
        print("Товар не знайдено.")

    def show_items(self):
        if not self.items:
            print("Кошик порожній.")
            return

        print("Товари у кошику:")
        for item in self.items:
            print(f"- {item.name}: {item.quantity} шт. по {item.price} грн")

cart = ShoppingCart()

item1 = Item("Сервопривід", 100, 1)
item2 = Item("Плата", 400, 2)

cart.add_item(item1)
cart.add_item(item2)

cart.show_items()

cart.remove_item("Сервопривід")

cart.show_items()
```

Ітерабельний об'єкт, це об'єкт, по якому можна пройтись циклом, наприклад список. Ітератор це спеціальний об’єкт, що створюється з ітерабельного та по черзі повертає його елементи по одному, запам'ятовуючи позицію

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class Week:

    def __init__(self):
        self.days = [
            "Понеділок",
            "Вівторок",
            "Середа",
            "Четвер",
            "П’ятниця",
            "Субота",
            "Неділя"
        ]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.days):
            day = self.days[self.index]
            self.index += 1
            return day
        else:
            raise StopIteration
week = Week()


for day in week:
    print(day)
```
