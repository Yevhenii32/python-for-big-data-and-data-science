19. Створіть дві множини. Продемонстуйте над ними операції: обʼєднання, різниці, пересічі та симетричної різниці. Використовуйте методи, що не змінюють множини, а створюють нові.

```python
first_set = {1, 3, 5, 7}
second_set = {3, 4, 6, 8}
union_set = first_set.union(second_set)
diff_set = first_set.difference(second_set)
intersection_set = first_set.intersection(second_set)
symmetric_diff_set = first_set.symmetric_difference(second_set)
print("Об'єднання:", union_set)
print("Різниця:", diff_set)
print("Пересіч:", intersection_set)
print("Симетрична різниця:", symmetric_diff_set)
```
